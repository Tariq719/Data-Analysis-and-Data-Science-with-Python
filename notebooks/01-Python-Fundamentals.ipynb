#### **3. Variables and Data Types**

##### **3.1 What is a variable?**

A **variable** is a name that refers to a value stored by a program.
Variables allow us to keep and reuse information.

````python
student_name = "Muhammad Numan"
student_class = "9th"
obtained_marks = 1011
total_marks = 1200

percentage = obtained_marks / total_marks * 100

print(student_name)
print(student_class)
print(percentage)
````

**Output**

```text
Muhammad Numan
9th
84.25
```

##### **Variable naming rules**

-   A variable name may contain letters, digits, and underscores.
-   It cannot begin with a digit.
-   It cannot contain spaces.
-   Variable names are case-sensitive: `marks` and `Marks` are different
    names.
-   Use meaningful names such as `student_name` rather than unclear
    names such as `x` when possible.
-   Python keywords such as `if`, `for`, and `class` cannot be used as
    variable names.

#### **3.2 Common data types**

A **data type** describes the kind of value a variable refers to.

| Type | Description | Example |
|---|---|---|
| `int` | Whole numbers | `25`, `-3` |
| `float` | Numbers with a decimal part | `3.14`, `85.5` |
| `str` | Text | `"Python"` |
| `bool` | Logical values | `True`, `False` |
| `list` | Ordered, changeable collection | `[10, 20, 30]` |
| `tuple` | Ordered, unchangeable collection of elements | `(10, 20, 30)` |
| `set` | Collection of unique elements | `{10, 20, 30}` |
| `dict` | Key-value pairs | `{"name": "Ali"}` |

Use `type()` to inspect a value's type.

````python
name = "Ali"
student_class = "9th"
marks = 85
percentage = 85.5
is_passed = True

print(type(name))
print(type(student_class))
print(type(marks))
print(type(percentage))
print(type(is_passed))
````

**Output**

```text
<class 'str'>
<class 'str'>
<class 'int'>
<class 'float'>
<class 'bool'>
```

#### **3.3 Implicit type conversion**

Python sometimes converts a value from one numeric type to another
automatically.

````python
a = 5
b = 5.5
c = a + b

print(c)
print(type(c))
````

**Output**

```text
10.5
<class 'float'>
```

Here, Python converts the integer `5` to a floating-point value for the
calculation.

#### **4. Input and Type Conversion**

##### **4.1 Getting input from the user**

The `input()` function accepts information typed by a user. **It returns
the entered value as a string**, even if the user types digits.

````python
name = input("Enter your name: ")
student_class = input("Enter your class: ")

print("Name:", name)
print("Class:", student_class)
````

#### **4.2 Explicit type conversion**

Use conversion functions when a value needs another type.

| Function | Purpose | Example |
|---|---|---|
| `int()` | Convert a suitable value to an integer | `int("25")` → `25` |
| `float()` | Convert a suitable value to a float | `float("3.5")` → `3.5` |
| `str()` | Convert a value to text | `str(85)` → `"85"` |
| `bool()` | Convert a value to a Boolean | `bool(1)` → `True` |

##### *Example: Calculate a student's percentage*

````python
student_name = input("Enter student name: ")
obtained_marks = int(input("Enter obtained marks: "))
total_marks = int(input("Enter total marks: "))

if total_marks <= 0:
    print("Total marks must be greater than zero.")
else:
    percentage = obtained_marks / total_marks * 100
    print("Student:", student_name)
    print("Percentage:", percentage)
````

**Important:** `int(input(...))` works when the user enters a whole
number. If the user enters text that cannot be converted, Python raises
a `ValueError`. Input validation can be added to handle that situation.

#### **5. Arithmetic Operators**

Arithmetic operators perform mathematical calculations.

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `+` | Addition | `10 + 3` | `13` |
| `-` | Subtraction | `10 - 3` | `7` |
| `*` | Multiplication | `10 * 3` | `30` |
| `/` | Division | `10 / 3` | `3.333...` |
| `//` | Floor division | `10 // 3` | `3` |
| `%` | Remainder (modulo) | `10 % 3` | `1` |
| `**` | Exponentiation | `2 ** 3` | `8` |

##### *Example: Use arithmetic operators*

````python
value_1 = int(input("Enter the first whole number: "))
value_2 = int(input("Enter the second whole number: "))

print("Addition:", value_1 + value_2)
print("Subtraction:", value_1 - value_2)
print("Multiplication:", value_1 * value_2)

if value_2 != 0:
    print("Division:", value_1 / value_2)
    print("Floor division:", value_1 // value_2)
    print("Remainder:", value_1 % value_2)
else:
    print("Division, floor division, and remainder by zero are not allowed.")
````

**Remember:** `//` calculates floor division; `%` calculates the
remainder. They are not interchangeable.

#### **6. Assignment Operators**

Assignment operators store a value in a variable. Compound assignment
operators update the current value.

| Operator | Equivalent Form |
|---|---|
| `a += b` | `a = a + b` |
| `a -= b` | `a = a - b` |
| `a *= b` | `a = a * b` |
| `a /= b` | `a = a / b` |
| `a //= b` | `a = a // b` |
| `a %= b` | `a = a % b` |
| `a **= b` | `a = a ** b` |

##### *Example*

````python
a = 15
b = 3

a += b
print(a)  # 18

a -= b
print(a)  # 15

a *= b
print(a)  # 45

a /= b
print(a)  # 15.0
````

`/=` produces a floating-point result in Python 3, even when the result
is a whole number.

#### **7. Conditional Statements**

Conditional statements let a program choose which code to execute based
on whether a condition is true or false.

##### **7.1 `if`**

Run a block only when a condition is true.

````python
salary = 12000
expenses = 9000

if salary > expenses:
    print("Salary is greater than expenses.")
````

**Output**

```text
Salary is greater than expenses.
```

##### **7.2 `if-else`**

Choose between two paths.

````python
marks = 75

if marks >= 50:
    print("Pass")
else:
    print("Fail")
````

**Output**

```text
Pass
```

##### **7.3 `if-elif-else`**

Use multiple conditions when there are several possible outcomes.

````python
percentage = 84

if percentage >= 90:
    print("Grade: A+")
elif percentage >= 80:
    print("Grade: A")
elif percentage >= 70:
    print("Grade: B")
elif percentage >= 60:
    print("Grade: C")
elif percentage >= 50:
    print("Grade: D")
elif percentage >= 33:
    print("Grade: E")
else:
    print("Fail")
````

**Output**

```text
Grade: A
```

Python checks conditions from top to bottom and executes the first
matching block. These grade boundaries are an example for practice; use
the grading policy specified by your institution when assessing real
students.

#### **8. For Loops**

A **`for` loop** repeats a block of code for each item in an iterable,
such as a list, string, tuple, or range.

##### **8.1 Repeat a task using `range()`**

````python
for i in range(5):
    print("Hello")
````

This prints `Hello` five times. `range(5)` generates the integers `0`
through `4`.

##### **8.2 Print a sequence of numbers**

````python
for number in range(1, 6):
    print(number)
````

**Output**

```text
1
2
3
4
5
```

The stop value (`6`) is excluded from `range(1, 6)`.

##### **8.3 Loop through a list**

````python
students = ["Ali", "Ahmad", "Sara", "Ayesha"]

for student in students:
    print(student)
````

##### **8.4 Calculate total and average marks**

````python
marks = [75, 82, 91, 68, 88]

total = 0

for mark in marks:
    total += mark

average = total / len(marks)

print("Total:", total)
print("Average:", average)
````

**Output**

```text
Total: 404
Average: 80.8
```

`len(marks)` returns the number of marks in the list.

##### **8.5 Combine a loop with a condition**

````python
marks = [45, 78, 32, 90, 66]

for mark in marks:
    if mark >= 50:
        print(mark, "Pass")
    else:
        print(mark, "Fail")
````

##### **8.6 Print a multiplication table**

````python
number = int(input("Enter a number: "))

for multiplier in range(1, 11):
    print(number, "x", multiplier, "=", number * multiplier)
````

#### **9. While Loops**

A **`while` loop** repeats a block as long as its condition is true.

##### **9.1 Count from 1 to 5**

````python
count = 1

while count <= 5:
    print(count)
    count += 1
````

**Output**

```text
1
2
3
4
5
```

The variable is updated each time so the condition eventually becomes
false.

##### **9.2 Countdown**

````python
count = 5

while count >= 0:
    print(count)
    count -= 1

print("Start!")
````

**Output**

```text
5
4
3
2
1
0
Start!
```

##### **9.3 Repeat input until a positive number is entered**

````python
number = 0

while number <= 0:
    number = int(input("Enter a positive whole number: "))

print("You entered:", number)
````

The loop keeps asking while the number is zero or negative. This example
assumes the user enters valid whole-number text.

##### **9.4 Loop through a list using an index**

````python
numbers = [4, 6, 7, 9, 11, 16]
index = 0

while index < len(numbers):
    if numbers[index] % 2 == 0:
        print(numbers[index])
    index += 1
````

**Output**

```text
4
6
16
```

The index begins at `0` and increases by one each iteration. This
prevents an infinite loop.

### `for` versus `while`

-   Use a `for` loop when iterating over items or a known sequence.
-   Use a `while` loop when repetition depends on a condition that may
    change during execution.
-   Make sure a `while` loop can eventually end; otherwise, it may run
    forever.

#### **10. `break` and `continue`**

These statements change the normal flow of a loop.

-   **`break`** immediately exits the nearest enclosing loop.
-   **`continue`** skips the remaining statements in the current
    iteration and proceeds to the next iteration.

##### **10.1 `break` with a `for` loop**

````python
for number in range(1, 11):
    if number == 5:
        break
    print(number)
````

**Output**

```text
1
2
3
4
```

The loop stops when `number` becomes `5`.

##### **10.2 `continue` with a `for` loop**

````python
for number in range(1, 6):
    if number == 3:
        continue
    print(number)
````

**Output**

```text
1
2
4
5
```

Only the iteration for `3` is skipped.

##### **10.3 `break` with a `while` loop**

````python
number = 1

while number <= 10:
    if number == 5:
        break
    print(number)
    number += 1
````

**Output**

```text
1
2
3
4
```

##### **10.4 `continue` with a `while` loop**

````python
number = 0

while number < 10:
    number += 1

    if number in (5, 6, 7):
        continue

    print(number)
````

**Output**

```text
1
2
3
4
8
9
10
```

**Important:** In a `while` loop, update the loop-control variable
before a `continue` that would otherwise skip its update. If the
condition remains true and the variable never changes, the loop can
become infinite.

#### **11. Lists**

A **list** is an ordered, mutable collection that can store multiple
values. Lists can contain duplicate values and values of different
types.

````python
marks = [50, 45, 60, 35, 86, 79]
print(marks)
````

##### **11.1 Indexing**

List indexing starts at `0`.

````python
marks = [50, 45, 60, 35, 86, 79]

print(marks[0])   # First item: 50
print(marks[5])   # Last item: 79
print(marks[-1])  # Last item: 79
print(marks[-3])  # Third item from the end: 35
````

Positive indexes count from the start; negative indexes count from the
end.

##### **11.2 Slicing**

Slicing extracts a portion of a list. The stop index is excluded.

````python
marks = [50, 45, 60, 35, 86, 79]

print(marks[0:2])  # [50, 45]
print(marks[-3:-1])  # [35, 86]
print(marks[1:])   # [45, 60, 35, 86, 79]
print(marks[:4])   # [50, 45, 60, 35]
print(marks[:])    # A shallow copy of the list
print(marks[::2])  # [50, 60, 86]
````

General form: `list_name[start:stop:step]`.

##### **11.3 Update a list item**

Lists are mutable, so an item can be replaced by assigning to its index.

````python
marks = [50, 45, 60, 35]
marks[0] = 55

print(marks)
````

**Output**

```text
[55, 45, 60, 35]
```

##### **11.4 Add items: `append()`, `extend()`, and `insert()`**

````python
marks = [50, 45, 60]

marks.append(90)          # Adds one item at the end
marks.extend([92, 95])    # Adds each item from another iterable
marks.insert(1, 70)      # Inserts 70 at index 1

print(marks)
````

**Output**

```text
[50, 70, 45, 60, 90, 92, 95]
```

-   `append(item)` adds one item.
-   `extend(iterable)` adds each item from an iterable.
-   `insert(index, item)` inserts an item at a chosen index.

##### **11.5 Remove items: `pop()`, `remove()`, and `clear()`**

````python
marks = [50, 45, 60, 35, 86]

last_item = marks.pop()  # Removes and returns the last item
print("Removed:", last_item)
print(marks)

marks.pop(0)             # Removes the item at index 0
marks.remove(60)          # Removes the first matching value 60
print(marks)

marks.clear()             # Removes every item
print(marks)
````

**Important differences:**

-   `pop()` removes by index and returns the removed item. Without an
    index, it removes the last item.
-   `remove(value)` removes the first matching value. It raises
    `ValueError` if the value is not present.
-   `clear()` empties the list.

##### **11.6 Search and count: `index()` and `count()`**

````python
marks = [75, 90, 75, 60, 90]

print(marks.index(90))  # Index of the first 90: 1
print(marks.count(75))  # Number of occurrences of 75: 2
````

`index()` raises `ValueError` if the value is not found.

##### **11.7 Sort and reverse**

````python
marks = [75, 40, 90, 60]

marks.sort()
print(marks)  # [40, 60, 75, 90]

marks.sort(reverse=True)
print(marks)  # [90, 75, 60, 40]

marks.reverse()
print(marks)  # [40, 60, 75, 90]
````

-   `sort()` arranges list items in place.
-   `sort(reverse=True)` sorts in descending order.
-   `reverse()` reverses the current order; it does not sort the values.

##### **11.8 Copy a list**

````python
marks = [75, 80, 90]
marks_copy = marks.copy()

marks_copy.append(95)

print("Original:", marks)
print("Copy:", marks_copy)
````

**Output**

```text
Original: [75, 80, 90]
Copy: [75, 80, 90, 95]
```

`copy()` creates a shallow copy. For a list of simple values such as
numbers or strings, this is usually enough to create an independently
editable list.

#### **12. Dictionaries**

A **dictionary** stores information as **key-value pairs**. Keys are
unique, and values are accessed using their keys rather than numeric
indexes.

##### **12.1 Create and access a dictionary**

````python
student = {
    "Name": "Ahmad",
    "Father Name": "Junaid",
    "Roll No": 14
}

print(student["Name"])
print(student.get("Father Name"))
````

**Output**

```text
Ahmad
Junaid
```

-   `student["Name"]` accesses the value for the key `"Name"` and raises
    `KeyError` if the key is missing.
-   `student.get("Name")` accesses the value; if the key is missing, it
    returns `None` by default.

##### **12.2 Get keys, values, and items**

````python
student = {"Name": "Ahmad", "Roll No": 14, "Marks": 550}

print(student.keys())
print(student.values())
print(student.items())
````

-   `keys()` provides the dictionary's keys.
-   `values()` provides its values.
-   `items()` provides key-value pairs.

> Convert the values to a list if needed:

````python
values_list = list(student.values())
print(values_list)
````

##### **12.3 Add and update values**

````python
student = {"Name": "Ahmad", "Roll No": 14}

student["Marks"] = 550       # Adds a new key
student["Name"] = "Faisal"   # Updates an existing key
student.update({"Class": "9th", "Marks": 530})

print(student)
````

##### **12.4 Remove values with `pop()` and `del`**

````python
student = {"Name": "Ahmad", "Roll No": 14, "Marks": 550}

removed_marks = student.pop("Marks")
print("Removed marks:", removed_marks)
print(student)

del student["Roll No"]
print(student)
````

-   `pop(key)` removes the key and returns its value.
-   `del dictionary[key]` removes the key-value pair.
-   `pop()` raises `KeyError` if the key is missing and no default is
    provided.

##### **12.5 Loop through a dictionary**

````python
student = {"Name": "Ahmad", "Roll No": 14, "Marks": 550}

for key, value in student.items():
    print(key, ":", value)
````

**Output**

```text
Name : Ahmad
Roll No : 14
Marks : 550
```

##### **12.6 Move a value from a dictionary to a list**

````python
student = {"Name": "Ahmad", "Roll No": 14, "Marks": 550}
values = []

values.append(student.pop("Roll No"))

print(student)
print(values)
````

**Output**

```text
{'Name': 'Ahmad', 'Marks': 550}
[14]
```

`pop("Roll No")` removes the key and returns its value. `append()` then
adds that returned value to the list.

#### **14. The `return` Statement**

The `return` statement ends a function call and sends a value back to
the caller, if a value is specified. The returned value can be stored,
used in calculations, or passed to another function.

##### **14.1 Return a result**

````python
def add(a, b):
    return a + b

result = add(10, 20)
print(result)
````

**Output**

```text
30
```

##### **14.2 `print()` versus `return`**

````python
def add_with_print(a, b):
    print(a + b)

result = add_with_print(10, 20)
print("Returned value:", result)
````

**Output**

```text
30
Returned value: None
```

The function prints `30`, but because it has no explicit return value,
the function call returns `None`.

Compare it with:

````python
def add_with_return(a, b):
    return a + b

result = add_with_return(10, 20)
print(result)
print(result * 2)
````

**Output**

```text
30
60
```

##### **14.3 Return with a condition**

````python
def check_result(marks):
    if marks >= 50:
        return "Pass"
    return "Fail"

print(check_result(75))
print(check_result(40))
````

**Output**

```text
Pass
Fail
```

Once Python executes `return`, it immediately exits that function call.

##### **14.4 Return a list**

````python
def get_even_numbers(numbers):
    even_numbers = []

    for number in numbers:
        if number % 2 == 0:
            even_numbers.append(number)

    return even_numbers

numbers = [4, 7, 8, 11, 12, 15]
print(get_even_numbers(numbers))
````

**Output**

```text
[4, 8, 12]
```

#### **16. Sets**

A **set** is a collection of unique elements. Duplicate values are
removed automatically. Sets do not support indexing, and their iteration
order is not guaranteed.

##### **16.1 Remove duplicate values**

````python
numbers = {10, 20, 10, 30, 20, 40}
print(numbers)
````

The set contains the unique values `10`, `20`, `30`, and `40`. Do not
rely on a particular display order.

You can also remove duplicates from a list:

````python
numbers = [10, 20, 10, 30, 20, 40]
unique_numbers = set(numbers)

print(unique_numbers)
````

##### **16.2 Add and remove elements**

````python
fruits = {"Apple", "Banana", "Mango"}

fruits.add("Orange")
fruits.remove("Banana")

print(fruits)
````

-   `add(value)` adds an element.
-   `remove(value)` removes an element and raises `KeyError` if it is
    not present.
-   `discard(value)` also removes an element but does not raise an error
    if it is missing.

##### **16.3 Set operations**

````python
class_A = {"Ali", "Ahmad", "Usman"}
class_B = {"Ahmad", "Bilal", "Usman"}

print("Union:", class_A.union(class_B))
print("Intersection:", class_A.intersection(class_B))
print("Difference A - B:", class_A.difference(class_B))
````

-   **Union:** All unique elements from both sets.
-   **Intersection:** Elements shared by both sets.
-   **Difference:** Elements in the first set that are not in the second
    set.

#### **17. File Handling**

**File handling** allows a program to create, read, and write files so
data can be stored beyond a single variable's lifetime.

The examples below use files in the current working directory. In a
notebook environment, the working directory may differ from the folder
where your notebook is saved.

##### **17.1 Common file modes**

| Mode | Purpose |
|---|---|
| `"r"` | Read an existing file. The file must exist. |
| `"w"` | Write to a file. Creates it if needed; **truncates/clears existing content** when opened. |
| `"a"` | Append to the end of a file. Creates it if needed; preserves existing content. |
| `"x"` | Create a new file; raises an error if the file already exists. |

Use `with open(...) as file:` to work with a file. The `with` statement
automatically closes the file when the block ends.

##### **17.2 Read an existing file --- `"r"`**

First make sure a file named `record.txt` exists in the working
directory and contains some text.

````python
with open("record.txt", "r", encoding="utf-8") as file:
    content = file.read()

print(content)
````

-   `open()` opens the file.
-   `"record.txt"` is the file path.
-   `"r"` means read mode.
-   `read()` reads the entire file content.
-   `encoding="utf-8"` supports a wide range of text characters.

If the file does not exist at that path, Python raises
`FileNotFoundError`.

##### **17.3 Write a student report --- `"w"`**

````python
student_name = "Ali"
marks = 85
grade = "A"

with open("student_report.txt", "w", encoding="utf-8") as file:
    file.write("Student Report\n")
    file.write("-----------------\n")
    file.write("Name: " + student_name + "\n")
    file.write("Marks: " + str(marks) + "\n")
    file.write("Grade: " + grade + "\n")

with open("student_report.txt", "r", encoding="utf-8") as file:
    print(file.read())
````

**Output**

```text
Student Report
-----------------
Name: Ali
Marks: 85
Grade: A
```

`write()` expects a string, so `str(marks)` converts the integer `85`
into text. Opening the same file with `"w"` again will replace its
previous contents.

##### **17.4 Replace the previous report --- `"w"` again**

````python
name = "Aisha"
subject = "Computer"
institute = "IT All Rounders"

with open("student_report.txt", "w", encoding="utf-8") as file:
    file.write("Marks Report\n")
    file.write("-----------------\n")
    file.write("Name: " + name + "\n")
    file.write("Subject: " + subject + "\n")
    file.write("Institute: " + institute + "\n")

with open("student_report.txt", "r", encoding="utf-8") as file:
    print(file.read())
````

The file now contains Aisha's report, not Ali's. This happens because
`"w"` clears the existing file content when it is opened.

**Key point:** It does not matter whether the code is run in the same
program or a different program. If the path points to the same file and
it is opened in `"w"` mode, its existing content is replaced.

##### **17.5 Append a teacher report --- `"a"`**

````python
teacher_name = "Salma"
course = "Data Analysis"
academy = "IT All Rounders"

with open("student_report.txt", "a", encoding="utf-8") as file:
    file.write("\nTeacher Report\n")
    file.write("-----------------\n")
    file.write("Teacher Name: " + teacher_name + "\n")
    file.write("Course: " + course + "\n")
    file.write("Academy: " + academy + "\n")

with open("student_report.txt", "r", encoding="utf-8") as file:
    print(file.read())
````

The file now contains Aisha's report followed by Salma's teacher report.
The `"a"` mode preserves the existing text and adds new text at the end.

##### **17.6 File handling summary**

-   Use `"r"` to read.
-   Use `"w"` to write or replace existing content.
-   Use `"a"` to add content without removing existing text.
-   Use `with` so the file is closed automatically.
-   Use a correct file path and handle errors when necessary.
